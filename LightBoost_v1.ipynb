## Limpieza

Decisiones tomadas por defecto (cámbialas si no encajan con el negocio):
- Se excluyen las columnas de fuga, **incluida `KG Base`** (no confirmado si se conoce antes del
  servicio; si sabes que sí, quítala de `leakage_cols` para probar con ella).
- Se excluyen filas de `plato` que no son platos reales (cambios de buffet, comedor del personal, cierres).
- Se eliminan filas sin `Comensales` (~2% de los datos).

```python
target = "Desperdicio (kg)"

leakage_cols = [
    "KG Reales", "CO2", "KG Evitados", "KG Evitados Totales",
    "% Reducción", "% Reducción Total", "kg/Comensal", "% KG", "Desviación %",
    "KG Base",
]
df = df.drop(columns=leakage_cols)

patron_no_plato = df["plato"].str.contains(
    "Comedor del personal|Cambio del día|Desperdicio Cierre|Cambio de|Cierre Buffet",
    case=False, na=False
)
df = df.loc[~patron_no_plato].copy()

df = df.dropna(subset=["Comensales"]).copy()

print("Shape tras limpieza:", df.shape)
```

## Feature engineering

```python
df["dia_semana"] = df["Date"].dt.dayofweek
df["mes"] = df["Date"].dt.month
df["es_finde"] = df["dia_semana"].isin([5, 6]).astype(int)
```

```python
# Lag y media móvil por plato+turno, usando solo datos PASADOS (sin fuga temporal)
df = df.sort_values(["plato", "turno", "Date"]).reset_index(drop=True)
grp = df.groupby(["plato", "turno"])[target]

df["lag_1"] = grp.shift(1)
df["media_movil_7"] = grp.transform(lambda s: s.shift(1).rolling(7, min_periods=1).mean())

# Para el primer registro de cada plato+turno no hay histórico: se rellena con la mediana global
df["lag_1"] = df["lag_1"].fillna(df[target].median())
df["media_movil_7"] = df["media_movil_7"].fillna(df[target].median())
```

```python
for col in ["turno", "plato"]:
    df[col] = df[col].astype("category")

df.head()
```

## Split temporal

Nada de split aleatorio: se entrena con el pasado y se valida con las fechas más recientes
(últimos ~15% de los días), simulando el escenario real de producción.

```python
df = df.sort_values("Date").reset_index(drop=True)

corte = df["Date"].quantile(0.85, interpolation="nearest")
train = df[df["Date"] <= corte]
test = df[df["Date"] > corte]

print(f"Corte: {corte.date()}")
print("Train:", train.shape, "Test:", test.shape)
```

```python
features = ["turno", "plato", "Comensales", "dia_semana", "mes", "es_finde", "lag_1", "media_movil_7"]

X_train, y_train = train[features], train[target]
X_test, y_test = test[features], test[target]

y_train_log = np.log1p(y_train)
y_test_log = np.log1p(y_test)
```

## Modelo baseline

Referencia mínima: media histórica de desperdicio por plato+turno (calculada solo con train).
El LightGBM tiene que batir esto para justificar la complejidad.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

baseline_map = train.groupby(["plato", "turno"], observed=True)[target].mean()
global_mean = train[target].mean()

test_keys = list(zip(test["plato"], test["turno"]))
baseline_pred = np.array([baseline_map.get(k, global_mean) for k in test_keys])

mae_base = mean_absolute_error(y_test, baseline_pred)
rmse_base = np.sqrt(mean_squared_error(y_test, baseline_pred))

print(f"Baseline -> MAE: {mae_base:.3f} kg | RMSE: {rmse_base:.3f} kg")
```

## LightGBM

Se entrena sobre `log1p(Desperdicio (kg))` por el sesgo visto en el EDA, y se deshace la
transformación (`expm1`) antes de calcular métricas, para que MAE/RMSE queden en kg reales.
`turno` y `plato` van como categóricas nativas de LightGBM (sin one-hot).

```python
import lightgbm as lgb

train_set = lgb.Dataset(X_train, label=y_train_log, categorical_feature=["turno", "plato"])
val_set = lgb.Dataset(X_test, label=y_test_log, categorical_feature=["turno", "plato"], reference=train_set)

params = {
    "objective": "regression",
    "metric": "mae",
    "learning_rate": 0.005,
    "num_leaves": 30,
    "verbose": -1,
}

model = lgb.train(
    params,
    train_set,
    num_boost_round=1000,
    valid_sets=[val_set],
    callbacks=[lgb.early_stopping(50), lgb.log_evaluation(50)],
)
```

## Métricas

```python
pred_log = model.predict(X_test, num_iteration=model.best_iteration)
pred = np.expm1(pred_log)
pred = np.clip(pred, 0, None)  # el desperdicio no puede ser negativo

mae = mean_absolute_error(y_test, pred)
rmse = np.sqrt(mean_squared_error(y_test, pred))

print(f"LightGBM -> MAE: {mae:.3f} kg | RMSE: {rmse:.3f} kg")
print(f"Baseline -> MAE: {mae_base:.3f} kg | RMSE: {rmse_base:.3f} kg")
print(f"Mejora MAE vs baseline: {(1 - mae/mae_base)*100:.1f}%")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, pred, alpha=0.3)
lims = [0, max(y_test.max(), pred.max())]
plt.plot(lims, lims, "r--")
plt.xlabel("Desperdicio real (kg)")
plt.ylabel("Desperdicio predicho (kg)")
plt.title("Real vs predicho")
plt.show()
```

```python
lgb.plot_importance(model, max_num_features=15, importance_type="gain", figsize=(8, 6))
plt.show()
```

```python
params_tweedie = {
    "objective": "tweedie",
    "tweedie_variance_power": 1.3,
    "metric": "mae",
    "learning_rate": 0.05,
    "num_leaves": 31,
    "min_data_in_leaf": 25,
    "verbose": -1,
}

train_set_raw = lgb.Dataset(X_train, label=y_train, categorical_feature=["turno", "plato"])
val_set_raw = lgb.Dataset(X_test, label=y_test, categorical_feature=["turno", "plato"], reference=train_set_raw)

model_tweedie = lgb.train(
    params_tweedie,
    train_set_raw,
    num_boost_round=1000,
    valid_sets=[val_set_raw],
    callbacks=[lgb.early_stopping(30), lgb.log_evaluation(50)],
)

pred_tweedie = model_tweedie.predict(X_test, num_iteration=model_tweedie.best_iteration)
pred_tweedie = np.clip(pred_tweedie, 0, None)

mae_tweedie = mean_absolute_error(y_test, pred_tweedie)
rmse_tweedie = np.sqrt(mean_squared_error(y_test, pred_tweedie))

print(f"Tweedie  -> MAE: {mae_tweedie:.3f} kg | RMSE: {rmse_tweedie:.3f} kg")
print(f"log1p    -> MAE: {mae:.3f} kg | RMSE: {rmse:.3f} kg")
print(f"Baseline -> MAE: {mae_base:.3f} kg | RMSE: {rmse_base:.3f} kg")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, pred_tweedie, alpha=0.3)
lims = [0, max(y_test.max(), pred_tweedie.max())]
plt.plot(lims, lims, "r--")
plt.xlabel("Desperdicio real (kg)")
plt.ylabel("Desperdicio predicho (kg) - tweedie")
plt.title("Real vs predicho (tweedie)")
plt.show()
```

```python
import xgboost as xgb

dtrain = xgb.DMatrix(X_train, label=y_train_log, enable_categorical=True)
dval = xgb.DMatrix(X_test, label=y_test_log, enable_categorical=True)

params_xgb = {
    "objective": "reg:squarederror",
    "eval_metric": "mae",
    "learning_rate": 0.05,
    "max_depth": 6,
    "min_child_weight": 5,
}

model_xgb = xgb.train(
    params_xgb,
    dtrain,
    num_boost_round=1000,
    evals=[(dval, "valid")],
    early_stopping_rounds=30,
    verbose_eval=50,
)
```

```python
pred_xgb_log = model_xgb.predict(dval, iteration_range=(0, model_xgb.best_iteration + 1))
pred_xgb = np.expm1(pred_xgb_log)
pred_xgb = np.clip(pred_xgb, 0, None)

mae_xgb = mean_absolute_error(y_test, pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test, pred_xgb))

print(f"XGBoost  -> MAE: {mae_xgb:.3f} kg | RMSE: {rmse_xgb:.3f} kg")
print(f"LightGBM -> MAE: {mae:.3f} kg | RMSE: {rmse:.3f} kg")
print(f"Baseline -> MAE: {mae_base:.3f} kg | RMSE: {rmse_base:.3f} kg")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, pred_xgb, alpha=0.3)
lims = [0, max(y_test.max(), pred_xgb.max())]
plt.plot(lims, lims, "r--")
plt.xlabel("Desperdicio real (kg)")
plt.ylabel("Desperdicio predicho (kg) - XGBoost")
plt.title("Real vs predicho (XGBoost)")
plt.show()
```

```python
from catboost import CatBoostRegressor, Pool

cat_features = ["turno", "plato"]

train_pool = Pool(X_train, label=y_train_log, cat_features=cat_features)
val_pool = Pool(X_test, label=y_test_log, cat_features=cat_features)

model_cb = CatBoostRegressor(
    iterations=1000,
    learning_rate=0.05,
    depth=6,
    loss_function="MAE",
    eval_metric="MAE",
    early_stopping_rounds=30,
    verbose=False,
)

model_cb.fit(train_pool, eval_set=val_pool)
```

```python
pred_cb_log = model_cb.predict(X_test)
pred_cb = np.expm1(pred_cb_log)
pred_cb = np.clip(pred_cb, 0, None)

mae_cb = mean_absolute_error(y_test, pred_cb)
rmse_cb = np.sqrt(mean_squared_error(y_test, pred_cb))

print(f"CatBoost -> MAE: {mae_cb:.3f} kg | RMSE: {rmse_cb:.3f} kg | mejor_iteración: {model_cb.get_best_iteration()}")
print(f"XGBoost  -> MAE: {mae_xgb:.3f} kg | RMSE: {rmse_xgb:.3f} kg")
print(f"LightGBM -> MAE: {mae:.3f} kg | RMSE: {rmse:.3f} kg")
print(f"Baseline -> MAE: {mae_base:.3f} kg | RMSE: {rmse_base:.3f} kg")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, pred_cb, alpha=0.3)
lims = [0, max(y_test.max(), pred_cb.max())]
plt.plot(lims, lims, "r--")
plt.xlabel("Desperdicio real (kg)")
plt.ylabel("Desperdicio predicho (kg) - CatBoost")
plt.title("Real vs predicho (CatBoost)")
plt.show()
```
